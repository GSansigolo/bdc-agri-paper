# Creating a 10 days Sentinel-2 Data Cube
<hr style="border:1px solid #0077b9;">

To create a 10-day Sentinel-2 surface reflectance data cube covering the year **2025**, execute the `smosaic mosaic` command with the following parameters:

```bash
!smosaic mosaic \
  --name S2 \
  --collection S2_L2A-1 \
  --mosaic-method lcf \
  --start-year 2025 \
  --start-month 1 \
  --start-day 1 \
  --end-year 2025 \
  --end-month 12 \
  --end-day 31 \
  --duration-days 10 \
  --bbox "-54.1585,-31.8788,-53.5868,-31.2459" \
  --proj_output 4326 \
  --profile crop_condition
